```python

# Instalacja wymaganych bibliotek (jeśli nie są zainstalowane)
!pip install opencv-python-headless pillow numpy matplotlib

# Importowanie bibliotek
import cv2
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from google.colab import files
from google.colab.patches import cv2_imshow

# Funkcja do wczytania zdjęcia
def load_image():
    # Wczytanie zdjęcia z dysku lokalnego
    uploaded = files.upload()
    image_path = list(uploaded.keys())[0]
    img = cv2.imread(image_path)
    return img

# Funkcja do analizy zdjęcia
def analyze_image(img):
    # Konwersja na odcienie szarości
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Wykrywanie krawędzi za pomocą algorytmu Canny
    edges = cv2.Canny(gray, 100, 200)

    # Obliczenie histogramu kolorów dla każdego kanału (B, G, R)
    colors = ('b', 'g', 'r')
    plt.figure(figsize=(10, 4))
    for i, color in enumerate(colors):
        hist = cv2.calcHist([img], [i], None, [256], [0, 256])
        plt.plot(hist, color=color)
        plt.xlim([0, 256])
    plt.title('Histogram kolorów')
    plt.xlabel('Wartość piksela')
    plt.ylabel('Liczba pikseli')
    plt.show()

    # Wyświetlenie oryginalnego obrazu i krawędzi
    print("Oryginalne zdjęcie:")
    cv2_imshow(img)
    print("Wykryte krawędzie:")
    cv2_imshow(edges)

    # Podstawowe informacje o zdjęciu
    height, width, channels = img.shape
    print(f"Rozmiar zdjęcia: {width}x{height} pikseli")
    print(f"Liczba kanałów: {channels}")
    print(f"Średnia wartość pikseli (RGB): {np.mean(img, axis=(0,1))}")

# Główna funkcja
def main():
    print("Prześlij zdjęcie do analizy:")
    img = load_image()
    analyze_image(img)

# Uruchomienie programu
if __name__ == "__main__":
    main()

f"Liczba wykrytych twarzy: {len(faces)}"


```
